```python
import numpy as np
import nibabel as nib
from pathlib import Path
from scipy.spatial import cKDTree
```

```python
vx_loc, vx_adj, surf_norm = nib.load(Path("/home/common/bonaiuto/cued_action_meg/derivatives/processed/sub-001/surf/pial.ds.link_vector.fixed.gii")).agg_data()
```

```python
# Build KD-tree for efficient neighbor search
tree = cKDTree(vx_loc)

# Query neighbors within 5mm for each vertex
neighbors = tree.query_ball_tree(tree, r=5.0)

# Compute average neighborhood size (excluding the vertex itself)
avg_neighbors = np.mean([len(n) - 1 for n in neighbors])

print(f"Average number of vertices within 5mm: {avg_neighbors:.10f}")
```
